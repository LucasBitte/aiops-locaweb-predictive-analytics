```python
import pandas as pd
import numpy as np
import os
```

```python
# 1. Definir caminhos (subindo um nível para sair de 'notebooks')
BASE_DIR = os.path.dirname(os.getcwd())
INPUT_PATH = os.path.join(BASE_DIR, "data", "raw", "LW-DATASET.xlsx")

# 2. Carregar o arquivo original
print(f"Lendo: {INPUT_PATH}")
df = pd.read_excel(INPUT_PATH)
```

```python

print(df.dtypes)
```

```python

print("🔍 Inspeção da coluna: Entrou para KPI?")
print(df['Entrou para KPI?'].value_counts(dropna=False))

print("\n🔍 Inspeção da coluna: KPI Violado?")
print(df['KPI Violado?'].value_counts(dropna=False))
```

```python
object_cols = df.select_dtypes(include=['object']).columns
df[object_cols] = df[object_cols].astype("string")
```
